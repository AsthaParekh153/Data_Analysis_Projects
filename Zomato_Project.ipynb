##Charts:

#Q1) What is the distribution if restaurant ratings?


```python
sns.histplot(data=df,x='rating',color='orange',kde=True)
plt.title("Distribution of Restuarant Rating")
plt.show()
```

Most restaurant ratings between 3.3 and 4.0, showing that most restaurant have good customer ratings.

#Q.2) Which restaurant type is the most common?


```python
sns.countplot(data=df,x='listed_in')
plt.title("Restaurant type by number")
plt.show()
```

Dining is most common restaurant type, with highest number of restaurant. and cafes are the second most common type.

Q.3) Does online ordering affect restaurant ratings?


```python
sns.barplot(data=df,x='online_order',y='rating')
plt.title("Rating by Online order availability")
plt.show()
```

Restaurant with online ordering have a higher average rating than restaurant without online ordering. Online ordering is highest rating.

Q.4) Does  table booking affect restaurant rating?


```python
sns.barplot(data=df,x='book_table', y='rating')
plt.title("Rating by Table booking ")
plt.show()
```

Restaurant with table booking have a higher average rating than restaurant without table booking .Table booking is highest rating.

Q.5) Which restaurant type has highest average cose for two people?

```python
sns.barplot(data=df,x='listed_in',y='approx_cost')
plt.title('Cost by Restaurant type')
plt.xticks(rotation=45)
plt.show()
```

Buffet and other restaurant types have highest approx cost for two  people, while dining has lowest approx cost.

Q.6) Which are the top 10 restaurant with the highest number of votes?



```python
sns.barplot(data=df.sort_values(by='votes',ascending=False).head(10),
            x='votes',y='name')
plt.title('Top 10 restaurant By votes')
plt.show()
```

Empire restaurant has highest number of votes, followed by Meghana foods and Onesta.

Q.7) Is there a relationship between restaurant ratings and votes?

```python
sns.scatterplot(data=df, x='votes', y ='rating')
plt.title("Votes and Resataurant Ratings")
plt.show()
```

The chart shows the restaurants with higher votes usually have good rating,but the relationship is not the same for all restaurants.

Q.8) Which restaurant type has the highest average rating?

```python
sns.barplot(data=df,x='listed_in',y='rating')
plt.title("Average rating by Restaurant")
plt.show()
```

The Other restaurant type has the highest average rating, followed by buffet and cafes. Dining has lowest rating.

```python
avg_cost = df.groupby(
    'listed_in'
)['approx_cost'].mean().reset_index()

sns.barplot(
    data=avg_cost,
    x='listed_in',
    y='approx_cost'
)

plt.title('Average Cost by Restaurant Type')
plt.show()
```
