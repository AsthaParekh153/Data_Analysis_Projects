# Q1.Find the average house price (MEDV).

```python
df['MEDV'].mean()
```

The average house price value is approximately 22.53.

#Q2.Find the median house price.

```python
df['MEDV'].median()
```

The median gives the middle house-price value and is useful for understanding the typical price without being strongly affected by extreme values.

#Q3. Find the minimum and maximum house price.

```python
print("Minimum" , df['MEDV'].min())
print("Maximum:", df['MEDV'].max())
```

The minimum MEDV is 5.0, while the maximum is 50.0.

#Q.4 Find the average number of rooms (RM).

```python
df['RM'].mean()
```

The average number of rooms is approximately 6.28.

#Q.5 Find houses having more than 7 rooms.

```python
df[df['RM'] > 7]
```

This returns all houses where the average number of rooms is greater than 7.

#Q.6 Find the total number of houses where the house price (MEDV) is greater than 30.

```python
df[df['MEDV'] > 30].shape[0]
```

This gives the number of houses whose MEDV value is greater than 30.

#Q.7 Find the standard deviation of house prices.

```python
df['MEDV'].std()
```

The standard deviation shows how much house prices vary from their average value

#Q.8 Find the correlation between RM and MEDV.

```python
df['RM'].corr(df['MEDV'])
```

The correlation is positive, showing that houses with more rooms generally tend to have higher prices.

#Q.9 Find the correlation of all variables with MEDV.

```python
df.corr(numeric_only=True)['MEDV'].sort_values(ascending=False)
```

This helps identify which variables have the strongest positive and negative relationships with house prices. RM has a strong positive relationship, while LSTAT has a strong negative relationship.

#Q.10 Find the total number of houses having more than 6 rooms.

```python
df[df['RM'] > 6].shape[0]
```

This tells us how many houses in the dataset have an average of more than 6 rooms.

#Charts:

# Q.1 Plot the distribution of house prices using a histogram.

```python
plt.hist(df['MEDV'], bins=20)
plt.xlabel('House Price')
plt.ylabel('Number of Houses')
plt.title('Distribution of House Prices')
plt.show()
```

The histogram shows that house prices are distributed across different price ranges, with many observations concentrated around the middle range.

# Q.2 Create a scatter plot between rooms and house prices.

```python
plt.scatter(df['RM'], df['MEDV'])
plt.xlabel('Average Number of Rooms')
plt.ylabel('House Price')
plt.title('Rooms vs House Price')
plt.show()
```

The chart shows a positive relationship. Generally, houses with more rooms have higher house prices.

#Q.3 Create a scatter plot between LSTAT and MEDV.

```python
plt.scatter(df['LSTAT'], df['MEDV'])
plt.xlabel('LSTAT')
plt.ylabel('House Price')
plt.title('LSTAT vs House Price')
plt.show()
```

The chart shows a negative relationship. As LSTAT increases, house prices generally decrease.




#Q.4 Create a line chart for the first 30 house prices.

```python
plt.plot(df['MEDV'].head(30))
plt.xlabel('House Index')
plt.ylabel('House Price')
plt.title('First 30 House Prices')
plt.show()
```

The line chart shows that house prices vary considerably across the first 30 observations.

#Q.5 Create a boxplot for house prices.

```python
sns.boxplot(x=df['MEDV'])
plt.xlabel('House Price')
plt.title('Boxplot of House Prices')
plt.show()
```

The boxplot shows the spread, median and possible outliers in house prices.

#Q.6 Create a histogram of the average number of rooms (RM).

```python
plt.hist(df['RM'], bins=15)
plt.xlabel('Average Number of Rooms')
plt.ylabel('Number of Houses')
plt.title('Distribution of Number of Rooms')
plt.show()
```

The histogram shows the distribution of the average number of rooms. Most houses have a room count around the middle range, while very high or very low values occur less frequently.

#Q.7 Create a scatter plot between DIS and MEDV.

```python
plt.scatter(df['DIS'], df['MEDV'])
plt.xlabel('Distance to Employment Centres')
plt.ylabel('House Price')
plt.title('DIS vs House Price')
plt.show()
```

The scatter plot shows the relationship between distance to employment centres and house prices. The points are spread across the chart, indicating that the relationship is not very strong.

#Q.8 Create a boxplot of house prices based on CHAS.

```python
sns.boxplot(x='CHAS', y='MEDV', data=df)
plt.xlabel('Charles River')
plt.ylabel('House Price')
plt.title('House Price by CHAS')
plt.show()
```

The chart compares house prices for properties located near the Charles River and properties that are not near it.

#Q.9 Create a bar chart showing the average MEDV for CHAS categories.

```python
avg_price = df.groupby('CHAS')['MEDV'].mean()
plt.bar(avg_price.index, avg_price.values)
plt.xlabel('CHAS')
plt.ylabel('Average House Price')
plt.title('Average House Price by CHAS')
plt.show()
```

The bar chart compares the average house prices for properties based on their CHAS category. It helps us see whether properties near the Charles River have different average prices.

```python
sns.boxplot(x=df['RM'])
plt.xlabel('Average Number of Rooms')
plt.title('Distribution of Average Number of Rooms')
plt.show()
```
