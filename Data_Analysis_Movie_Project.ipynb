##Exploration with Visuals and Conclusions

**Q.1 Does higher budget mean higher popularity ? Is there a coefficent relationship ?**

```python
#Plot the relation between budget and popularity
x=df['budget']
y=df['popularity']
plt.scatter(x,y)
plt.title("Popularity by budget")
plt.xlabel('budget')
plt.ylabel('Popularity')
```

Higher budget doesn't always higher popularity.The graph shows weak relationship between budget and popularity.

```python
#As per median budget value,value divide into two groups : low and high
a=df['budget'].median()
low_budget=df.query('budget < {}'.format(a))
high_budget=df.query('budget >= {}'.format(a))
```

```python
#low budget and high budget mean values respecively
mean_popularity_of_low_budget = low_budget['popularity'].mean()
mean_popularity_of_high_budget = high_budget['popularity'].mean()
```

```python
#Bar chart with values as per above
locations=[1,2]
heights=[mean_popularity_of_low_budget,mean_popularity_of_high_budget]
labels = ['low','high']
plt.bar(locations,heights,tick_label=labels)
plt.title("Average popularity by budget")
plt.xlabel('Budget')
plt.ylabel('Average popularity')
plt.show()
```

```python
increase_per = (mean_popularity_of_high_budget - mean_popularity_of_low_budget)/ mean_popularity_of_high_budget *100
increase_per
```

The popularity if high- budget movies is higher than low-budget movie. The relationship is not very strong.

**Q.2 What length will recieve the highest popularity?**

```python
# <60 min: Short, 60min <= <= -120 min :Medium , >120min:Long
Short = df.query('runtime < {}'.format(100))
Medium = df.query('runtime < {}'.format(200))
Long = df.query('runtime > {}'.format(200))
```

```python
mean_popularity_short = Short['popularity'].mean()
mean_popularity_medium = Medium['popularity'].mean()
mean_popularity_long = Long['popularity'].mean()
```

```python
location = [1,2,3]
heights = [mean_popularity_short,mean_popularity_medium,mean_popularity_long]
labels=['Low','Medium','High']
plt.bar(location,heights,tick_label = labels)
plt.title('Popularity by Runtime')
plt.xlabel('Runtime')
plt.ylabel('Popularity')
plt.show()
```

Movie should not be too short or too long. Medium length is good for more popularity.

```python
#Plott the realtion between runtime and popularity
x=df['runtime']
y=df['popularity']
plt.scatter(x,y)
plt.title('Popularity by Runtime')
plt.xlabel('Runtime')
plt.ylabel('Popularity')
```

**Q.5 Which genres are most popular from year to year?**

```python
df.head()
```

```python
#Counts for per category.
def extract_data(column_name):
   data= df[column_name].str.cat(sep = '|')
   #Create panda series and store value separate
   data = pd.Series(data.split('|'))  #Split the generes by |
   #Display value in descending order
   count= data.value_counts(ascending= False) #Count the occurrence of each genres
   return count
```

```python
#Above function is for split genres and count the occurrence of each one
genre_count = extract_data('genres')
```

```python
genre_count
```

```python
#Create sperate dataframe for plot
df_genre_count = pd.DataFrame({'genres': genre_count.index, 'count':genre_count.values})
df_genre_count
```

```python
f,ax = plt.subplots(figsize= (23,9))
#use above dataframe
sns.barplot(x='count',y='genres',data=df_genre_count)
ax.set_title('Occurence per genre')
ax.set_xlabel('Occurrence')
ax.set_ylabel('Genres')
plt.show()
```
